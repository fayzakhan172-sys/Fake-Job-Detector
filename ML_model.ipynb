### 2. Combine the text columns into one column called "text"

```python
df["text"] = (
    df["title"].fillna("") + " "
    + df["company_profile"].fillna("") + " "
    + df["description"].fillna("") + " "
    + df["requirements"].fillna("")
)

df.drop(columns =['job_id','salary_range', 'industry', 'function', 'benefits', 'location', 'department','title', 'company_profile', 'description', 'requirements'], inplace = True)
```
